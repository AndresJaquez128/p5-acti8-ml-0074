```python
# gasto diario en un diccionario key 
# values lista de dias lista gastos
datos = {
    "Dia": ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"],
    "Temperatura": [12, 23, 13, 33, 29]
}

df = pd.DataFrame(datos)

print("=== TEMPERATURAS ===")
print(df)
```

```python
fila = df["Temperatura"].idxmax()

print("Día con mayor temperatura:", df.loc[fila, "Dia"])
print("Temperatura máxima:", df["Temperatura"].max(), "°C")
print("+-+-+-+-+-+Jupyter Andres-+-+-+-+-+-+-")
print("+-+-+-+-+-+Jaquez Camacho NC 0074-+-+-+-+-+-+-")
```
