```python
# gasto diario en un diccionario key value
# values lista de dias lista gastos

datos = {
    "Producto": ["Libro", "Pegamento", "autocorrector", "pluma", "mapamundi"],
    "Precio": [53, 16, 20, 18, 5],
    "Cantidad": [10, 76, 9, 105, 20]
}

df = pd.DataFrame(datos)

#OPERACION PARA EL TOTAL
df["Total"] = df["Precio"] * df["Cantidad"] 
print("=== VENTAS DE PAPELERÍA ===")
print(df)
```
